```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Load Dataset
df = pd.read_csv('Unemployment_Rate_upto_11_2020.csv')

# Column names se extra spaces remove karna
df.columns = df.columns.str.strip()

# 2. State-wise Average Unemployment Rate
state_unemp = df.groupby('Region')['Estimated Unemployment Rate (%)'].mean().reset_index()
state_unemp = state_unemp.sort_values(by='Estimated Unemployment Rate (%)', ascending=False)

print("Top 5 States with Highest Unemployment Rate:")
print(state_unemp.head())

# 3. Visualization using Matplotlib only
plt.figure(figsize=(12, 8))
plt.barh(state_unemp['Region'], state_unemp['Estimated Unemployment Rate (%)'], color='skyblue', edgecolor='black')
plt.title('Average Unemployment Rate by State in India (2020)', fontsize=14, fontweight='bold')
plt.xlabel('Estimated Unemployment Rate (%)', fontsize=12)
plt.ylabel('State / Region', fontsize=12)
plt.gca().invert_yaxis() # Top pe highest state dikhane ke liye
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()

# Save plot for GitHub
plt.savefig('unemployment_analysis_plot.png')
plt.show()
```
